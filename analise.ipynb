## Evolução dos Preços ##

```python
df_lineplot = pd.read_csv(r"C:\main\preco-combustivel-brasil\combustiveis-brasil.csv")

df_lineplot["referencia"] = pd.to_datetime(df_lineplot["referencia"])

price_columns_lineplot = [
    "referencia",
    "gasolina_comum_preco_revenda_avg",
    "etanol_hidratado_preco_revenda_avg",
    "oleo_diesel_preco_revenda_avg",
    "gas_cozinha_glp_preco_revenda_avg",
    "gas_natural_veicular_gnv_preco_revenda_avg"
]

df_filtered_lineplot = df_lineplot[price_columns_lineplot]

df_melted_lineplot = df_filtered_lineplot.melt(id_vars=["referencia"], var_name="Combustível", value_name="Preço Médio")

fuel_names_map_lineplot = {
    "gasolina_comum_preco_revenda_avg": "Gasolina Comum",
    "etanol_hidratado_preco_revenda_avg": "Etanol Hidratado",
    "oleo_diesel_preco_revenda_avg": "Óleo Diesel",
    "gas_cozinha_glp_preco_revenda_avg": "Gás de Cozinha (GLP)",
    "gas_natural_veicular_gnv_preco_revenda_avg": "Gás Natural Veicular (GNV)"
}
df_melted_lineplot["Combustível"] = df_melted_lineplot["Combustível"].map(fuel_names_map_lineplot)

fig_lineplot = px.line(df_melted_lineplot, x="referencia", y="Preço Médio", color="Combustível",
              title="Evolução Interativa do Preço Médio dos Combustíveis no Brasil",
              labels={
                  "referencia": "Data",
                  "Preço Médio": "Preço Médio (R$)"
              })


fig_lineplot.update_layout(
    hovermode="x unified", 
    xaxis_rangeslider_visible=True 
)


fig_lineplot.show()
```

## BoxPlot ##

```python
valores_colunas = [
    'gasolina_comum_preco_revenda_avg',
    'etanol_hidratado_preco_revenda_avg',
    'oleo_diesel_preco_revenda_avg',
    'gas_cozinha_glp_preco_revenda_avg',
    'gas_natural_veicular_gnv_preco_revenda_avg'
]

df_valores = df[valores_colunas]

df_boxplot = df_valores.melt(var_name='Combustível', value_name='Preço Médio')

renomear_colunas = {
    'gasolina_comum_preco_revenda_avg': 'Gasolina Comum',
    'etanol_hidratado_preco_revenda_avg': 'Etanol Hidratado',
    'oleo_diesel_preco_revenda_avg': 'Óleo Diesel',
    'gas_cozinha_glp_preco_revenda_avg': 'Gás de Cozinha (GLP)',
    'gas_natural_veicular_gnv_preco_revenda_avg': 'Gás Natural Veicular (GNV)'
}
df_boxplot['Combustível'] = df_boxplot['Combustível'].map(renomear_colunas)


fig = px.box(df_boxplot, x='Combustível', y='Preço Médio',
             title='Distribuição Interativa dos Preços Médios de Combustíveis no Brasil',
             labels={'Preço Médio': 'Preço Médio (R$)'},
             color='Combustível')


fig.update_layout(
    xaxis_title='Tipo de Combustível',
    yaxis_title='Preço Médio (R$)',
    hovermode='x unified'
)

fig.show()
```
